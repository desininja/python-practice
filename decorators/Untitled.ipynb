```python
def repeat(n):
    def decorator(func):
        def wrapper(a):
            for i in range(n):
                func(a)
        return wrapper
    return decorator
```

```python
@repeat(7)
def say_hello(a):
    print(f"Hello! {a}")
```

## Both *args and **kwargs are special syntaxes in Python used in function definitions to allow a function to accept a variable number of arguments.

#### *args (Non-Keyword Arguments)

*args is used to pass a non-keyworded, varibale length argument list to a function.
    - Syntax: The asterisk(`*`) is the key  part; `args` is just a conventional name (you could use `*data` or  `*items`).
    - Data Type: inside the function, args is treated as a tuple of all the arguments passed to the function after the required parameters.


```python
def sum_all(*args):
    print(f"Type of argument: {type(args)}")
    return sum(args)

total = sum_all(10,20,30)
print(f"Total: {total}")
```

```python
total_2 = sum_all(10,20,30,40,50)
print(f"Total: {total_2}")
```

```python
def some_operation(*items):
    print(f"Type of argument: {type(items)}")
    for i in items:
        print(i)
    return

some_operation('hi',1,'hello',4.5,{"pop":90},[1,2,"a"])
```

### **kwargs (Keyword Arguments)

** kwargs is used to pass a keyworded, variable-length argument list to a function. 
    - Syntax: the double-asterisk (`**`) is the key part; `kwargs` is the conventional name (short for "keyword arguments").
    - Data Type: Inside the function, `kwargs` is treated as a dictionary where the keys are the argument names and the values are the argument values.

```python
def access_info(**kwargs):

    print(f"Type of the 'kwargs', {type(kwargs)}")
    print(f"Name: {kwargs['name']}")
    print(f"Age: {kwargs['age']}")
    print(f"Profession: {kwargs['profession']}")
    print(len(kwargs))

    for key, value in kwargs.items():
        print(f"{key.capitalize()}: {value}")
    return 




access_info(name="Himanshu Bhatt",age= 29,profession="Data Engineer")
          

    
```

# Combining *args and **kwargs

It is common to use both in the same function definition to accept any kind of argument. If you use both, the order must be:
    -1 Standard arguments
    -2 `*args`
    -3 `**kwargs`

This combination is most often used when creating decorators or wrapper functions to ensure they can accept and correclty pass along all arguments to the original function they are wrapping.

```python
def greet_and_log(greeting1, greeting2, *people, **details):
    print(f"{greeting1} to all!")
    print(f"{greeting2} to some!")

    if people:
        print(f"People greeted: {people}")
    if details:
        print("Additional details:")
        for key, value in details.items():
            print(f"- {key}: {value}")

greet_and_log(
    "Welcome",    # Standard Argument
    "Hello",      #Standard Argument
    "John", "Jane","Himanshu",   #*peopel Tuple
    day="Sunday", time="7:22pm"  #**details (dictionary)
)
```

💥 Unpacking with * (For Positional Arguments)
The single asterisk (*) is used to unpack an iterable (like a list or tuple) into positional arguments when calling a function.

```python
def multiply(a, b, c):
    return a * b * c

my_list = [2, 3, 4]

# Unpacking the list into arguments: 
# It's as if you called multiply(2, 3, 4)
result = multiply(*my_list) 

print(f"Result: {result}")
```

🔑 Unpacking with ** (For Keyword Arguments)
The double asterisk (**) is used to unpack a dictionary into keyword arguments when calling a function. The keys of the dictionary must match the parameter names of the function.

```python
def access_info(name, age,profession):

    
    
    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"Profession: {profession}")

    return 


info ={"name":"Himanshu Bhatt","age": 29,"profession":"Data Engineer"}

access_info(**info)
```

Context,Syntax,Purpose,Data Type Handled
Defining a function,"def func(*args, **kwargs):",Collects multiple arguments into one variable.,*args becomes a tuple; **kwargs becomes a dictionary.
Calling a function,"func(*list, **dict)",Expands an iterable/dictionary into separate arguments.,* expands a list/tuple; ** expands a dictionary.

    This unpacking feature is also sometimes called the "splat operator" and is a key concept in Python for passing dynamic data structures to functions.

Decorator Handling *args and **kwargs

```python
def log_arguments_decorator(func):
    """
    A decorator that logs the arguments passed to the decorated function.
    """
    def wrapper(*args, **kwargs):
        
        # 1. Log arguments BEFORE the function call
        print(f"\n[LOG] Calling function: {func.__name__}")
        if args:
            print(f"[LOG] Positional arguments (*args): {args}")
        if kwargs:
            print(f"[LOG] Keyword arguments (**kwargs): {kwargs}")

        # 2. Execute the original function with the arguments
        #    The * and ** here are for UNPACKING the arguments
        result = func(*args, **kwargs)

        # 3. Log result AFTER the function call
        print(f"[LOG] Function {func.__name__} returned: {result}")
        
        # 4. Return the result
        return result
    return wrapper

# Apply the decorator to a function that takes arguments
@log_arguments_decorator
def calculate_power(base, exponent, verbose=True):
    """Calculates a number raised to a power."""
    if verbose:
        print(f"  --> Calculating {base} raised to {exponent}...")
    return base ** exponent

# --- Test Cases ---

# Case 1: Positional Arguments only
print("--- Test Case 1: Positional ---")
output_1 = calculate_power(2, 10)

# Case 2: Mixed Positional and Keyword Arguments
print("\n--- Test Case 2: Mixed ---")
output_2 = calculate_power(5, 3, verbose=False)
```
